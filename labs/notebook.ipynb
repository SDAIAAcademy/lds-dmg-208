## الإعداد

في Colab تنسخ الخلية المستودع وتثبّت المكتبات (دقيقة أو دقيقتان). على حاسوبك افتح الدفتر من مجلد `labs` وتجاوز النسخ.

```python
import os, subprocess, sys
REPO = "https://github.com/SDAIAAcademy/lds-dmg-208"
if "google.colab" in sys.modules and not os.path.exists("lds-dmg-208"):
    subprocess.run(["git", "clone", "--depth", "1", REPO], check=True)
if os.path.isdir("lds-dmg-208/labs"):
    os.chdir("lds-dmg-208/labs")
if "google.colab" in sys.modules:
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", "requirements-colab.txt"], check=True)
print("working in", os.getcwd())
```

```python
def kit(*args):
    """Run a course kit command, as `uv run kit/...` does on a laptop."""
    r = subprocess.run([sys.executable, *args], capture_output=True, text=True, encoding="utf-8")
    print(r.stdout[-6000:])
    if r.returncode:
        print(r.stderr[-3000:])
```
